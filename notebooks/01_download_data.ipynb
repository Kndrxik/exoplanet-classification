```python
from urllib.parse import quote_plus

import pandas as pd


query = """
SELECT
    kepoi_name,
    koi_disposition,
    koi_period,
    koi_duration,
    koi_depth,
    koi_impact,
    koi_prad,
    koi_teq,
    koi_insol,
    koi_model_snr
FROM cumulative
WHERE koi_disposition IN ('CONFIRMED', 'FALSE POSITIVE')
"""

url = (
    "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?query="
    + quote_plus(" ".join(query.split()))
    + "&format=csv"
)

df = pd.read_csv(url)

print(df.shape)
print(df.head())
print(df["koi_disposition"].value_counts())

df.to_csv("kepler_exoplanet_classification.csv", index=False)
```
